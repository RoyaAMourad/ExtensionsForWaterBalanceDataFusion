```python
#useful imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import glob
from scipy.stats import norm#the normal distribution
from matplotlib.ticker import MultipleLocator, AutoMinorLocator
from matplotlib.font_manager import FontProperties
import seaborn as sns
from collections import OrderedDict
import datetime
import matplotlib.dates as mdates
from scipy.stats import norm
```

```python

# Load data (priors as txt files)
PObs1 = np.loadtxt("minP.txt")
PObs2 = np.loadtxt("maxP.txt")
PStd = np.loadtxt("gpmErrorC.txt")
EObs1 = np.loadtxt("minET.txt")
EObs2 = np.loadtxt("maxET.txt")
QObs = np.loadtxt("runoff.txt")
IObs = np.loadtxt("waterImports.txt")
SObs = np.loadtxt("graceMAS.txt")
IMD=np.loadtxt("IMD-P.txt")
etref_df = np.loadtxt(r'/Users/rmourad/Library/CloudStorage/OneDrive-DelftUniversityofTechnology/3.chp2/galeta_subbasin/pars_plotting/data/E/TC_ETref.txt')
etref= np.loadtxt(r'/Users/rmourad/Library/CloudStorage/OneDrive-DelftUniversityofTechnology/3.chp2/galeta_subbasin/pars_plotting/data/E/ERA5ETref.txt')

# Create df to hold priors
priorDict = { # prior dictionary
    "Pmin": PObs1,
    "Pmax": PObs2,
    "PStd": PStd,
    "Emin": EObs1,
    "Emax": EObs2,
    "QObs": QObs,
    "IObs": IObs,
    "SObs": SObs,
    "Ep":etref,
    "Ep2":etref_df,
    "IMD": IMD
}

dfPriors = pd.DataFrame(priorDict)

print(dfPriors.head())
```

```python
posteriors_files = [(file, open(file, 'r').read()) for file in glob.glob('*.out') if file != 'resultsPar.out'if file != 'resultsS0.out']

# parse posteriors files 
posteriors = {}
for file, data in posteriors_files:
    filename = file.split('.')[0]
    parsed_data = [tuple(map(float, entry.strip('Gaussian()').replace('(', '').replace(')', '').split(','))) for entry in data.split('\n') if entry]
    mean = [entry[0] for entry in parsed_data]
    variance = [entry[1] for entry in parsed_data]
    posteriors['mPost'+filename.split('results')[1]] = mean
    posteriors['vPost'+filename.split('results')[1] ] = variance

# Create df to hold posteriors
dfPosteriors = pd.DataFrame(posteriors)

print(dfPosteriors.head())
```

```python
dfPosteriors['residual']=dfPosteriors['mPostP']-dfPosteriors['mPostE']+dfPosteriors['mPostI']-dfPosteriors['mPostQ']
dfPosteriors['dmPostS']=dfPosteriors['mPostS'].diff()

dfPosteriors.to_excel('Els50KmPls50kmRescaled.xlsx')
```

```python
# merge dfs
wb = pd.concat([dfPriors, dfPosteriors], axis=1)
print(wb.head())
```

```python
dates = pd.date_range(start='1/1/2003', end='12/31/2022', freq='MS')
df_dates = pd.DataFrame({'date': dates})
#merge two dfs
df_merge = df_dates.combine_first(wb)
df_merge['IMD']=df_merge['IMD'].replace(-99999, np.nan)
df_merge
```

```python
def convert_col_to_date(df):
    df['date'] = pd.to_datetime(df[df["date"]])
    df = df.set_index(pd.to_datetime(df[df.columns[0]]))
    df.drop([df.columns[0], "date"],axis = 1, inplace = True)
    return df
```

```python
df_merge["date"]=pd.to_datetime(df_merge["date"])
df_merge=df_merge.set_index(df_merge["date"])

#df_merge=df_merge.drop(["date"],axis=1)
df_merge['month']=df_merge['date'].dt.month
df_merge
```

```python
def make_monthly_plots(df, coldict, var):
    '''
    for a dataframe of a single variable and a dict mapping col names to datasets, return:
    - the mean monthly scatterplot

    '''
    colors = sns.color_palette()
    cdict = OrderedDict(zip(list(coldict.values()),colors))

    # setup
    #fig = plt.figure(figsize = (19,9))
    fig, ax = plt.subplots(figsize =(4,3))
    
    ## Monthly mean plot 
    monthly = df.groupby(df.index.month).mean()
    
    for k,v in cdict.items():
        monthly[k].plot(ax=ax, marker='o', markersize=3, linestyle='-', x_compat=True, color=v)
    
    labels = [datetime.date(1900, x, 1).strftime('%B')[:3] for x in df.groupby(df.index.month).mean().index]
    ax.set_xticks(ticks = df.groupby(df.index.month).mean().index)
    ax.set_xticklabels(labels,rotation = 90, fontsize = 14)
    ax.legend(frameon=False, loc='best', fontsize=10)
    ax.set_title("Monthly Mean mPostE", size = 16)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import OrderedDict
import datetime
import math

def make_yearly_monthly_plots(df, coldict, var):
    '''
    For a dataframe of a single variable and a dict mapping col names to datasets, return:
    - a grid of mean monthly scatterplots, one for each year
    '''
    colors = sns.color_palette()
    cdict = OrderedDict(zip(list(coldict.values()), colors))
    
    # Get the range of years in the dataset
    years = df.index.year.unique()
    n_years = len(years)
    
    # Calculate the grid size
    n_cols = min(5, n_years)  # Max 5 columns
    n_rows = math.ceil(n_years / n_cols)
    
    # Setup
    fig, axs = plt.subplots(n_rows, n_cols, figsize=(4*n_cols, 3*n_rows), sharex=True, sharey=True)
    fig.suptitle(f"Monthly Mean {var} by Year", size=20)
    
    # Flatten axs if it's a 2D array
    if n_rows > 1:
        axs = axs.flatten()
    
    for idx, year in enumerate(years):
        ax = axs[idx] if n_rows > 1 else axs
        year_data = df[df.index.year == year]
        
        # Monthly mean plot for each year
        monthly = year_data.groupby(year_data.index.month).mean()
        
        for k, v in cdict.items():
            monthly[k].plot(ax=ax, marker='o', markersize=3, linestyle='-', x_compat=True, color=v, label=k)
        
        labels = [datetime.date(1900, x, 1).strftime('%B')[:3] for x in monthly.index]
        ax.set_xticks(monthly.index)
        ax.set_xticklabels(labels, rotation=90, fontsize=8)
        ax.set_title(f"{year}", size=12)
        
        # Only show y-label for leftmost subplots
        if idx % n_cols == 0:
            ax.set_ylabel("mm", size=10)
        
        # Only show legend for the first subplot
        if idx == 0:
            ax.legend(frameon=False, loc='best', fontsize=8)
    
    # Remove any unused subplots
    for idx in range(n_years, len(axs)):
        fig.delaxes(axs[idx])
    
    plt.tight_layout()
    
```

```python
def make_plots_post(df, S_STD=S_Std):
    '''
    for a dataframe of all priors and posteriors,
    return monthly full timeseries for each water balance variable
    '''
    # setup
    fig, axs = plt.subplots(5, 1, figsize=(12, 12.5), sharex=False)
    plt.rcParams['legend.fontsize'] = 12
    variables = ['P', 'E', 'Q', 'I', 'S'] #,
    wb_variables = ['Precipitation', 'Evaporation','River Discharge', 'Water Imports', 'Storage'] # 
    
    calibri_font = FontProperties(fname='Calibri Regular.ttf')
    
    for i, var in enumerate(variables):
        m = df['mPost' + var]
        v = df['vPost' + var]
        
        if var == 'P':
            obs1 = df[var + 'min']
            obs2 = df[var + 'max']
            PlotPosterior(m, v, obs1, obs2, axs[i], var)
            #imd_p = df['IMD'].replace(-99999, np.nan)
            #axs[i].plot(imd_p, '#6B9AC4', marker='.', markersize=8, linestyle='None', label='IMD-P')
            
            # Create a boolean mask for non-NaN values
            valid_mask = ~df['IMD'].isna()
    
            # Plot only valid IMD points
            axs[i].plot(np.arange(len(df))[valid_mask], 
                df['IMD'][valid_mask], 
                '#6B9AC4', marker='.', markersize=8, 
                linestyle='None', label='IMD-P')
            
        elif var == 'S':
            obs = df[var + 'Obs']
            PlotPosterior(m, v, obs, None, axs[i], var)
            # Add predictive posterior for GRACE
            PlotPredictivePosterior(df['Predicted_posterior'], S_STD**2, obs, axs[i])
        elif var == 'Q' or var == 'I':
            obs = df[var + 'Obs']
            PlotPosterior(m, v, obs, None, axs[i], var)
        elif var == 'E':
            obs1 = df[var + 'min']
            obs2 = df[var + 'max']
            PlotPosterior(m, v, obs1, obs2, axs[i], var)
        
        # Format y-label
        axs[i].set_ylabel(wb_variables[i] + ' (mm)', size=13)
        axs[i].yaxis.set_minor_locator(AutoMinorLocator())
        axs[i].tick_params(axis='y', which='major', labelsize=11, size=7)
        
        # Generate custom x-axis labels
        dates = pd.date_range(start='1/1/2003', end='31/12/2022', freq='YS')
        axs[i].set_xticks(np.arange(0, len(dates) * 12, 12))
        axs[i].set_xticklabels(dates.strftime('%Y'),rotation=45)
        axs[i].tick_params(axis='x', which='major', labelsize=11)
        
    plt.tight_layout()
    plt.style.use('default')
    plt.show()
    fig.savefig(f'timeseriesrunavg-predpost.jpg', dpi=300, bbox_inches='tight')

def PlotPosterior(m, v, obs1, obs2, ax, var):
    mu = m
    sigma = np.sqrt(v)
    lower = norm.ppf(0.05, mu, sigma)
    upper = norm.ppf(0.95, mu, sigma)
    fill = ax.fill_between(np.arange(len(lower)), lower, upper, color='grey', alpha=0.2, label=var, linewidth=1.75)
    if obs1 is not None:
        if obs2 is not None:
            # Case for P and E
            ax.plot(np.ma.masked_values(obs1, -99999), 'xkcd:dust', marker='.', markersize=8, linestyle='None', label=var + 'min')
            ax.plot(np.ma.masked_values(obs2, -99999), 'crimson', marker='.', markersize=8, linestyle='None', label=var + 'max')
        else:
            # Case for S, Q, and I
            ax.plot(np.ma.masked_values(obs1, -99999), 'xkcd:dust', marker='.', markersize=8, linestyle='None', label=var + 'Obs')
    ax.legend(loc='best', frameon=False)

def PlotPredictivePosterior(m, v, obs, ax):
    mu = m
    sigma = np.sqrt(v)
    lower = norm.ppf(0.05, mu, sigma)
    upper = norm.ppf(0.95, mu, sigma)
    ax.fill_between(np.arange(len(lower)), lower, upper, color='green', alpha=0.3, label='PSobs', linewidth=1.75)
    ax.legend(loc='best', frameon=False)
```
